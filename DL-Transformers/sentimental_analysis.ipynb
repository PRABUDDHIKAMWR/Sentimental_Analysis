```python
from transformers import AutoTokenizer
from transformers import AutoModelForSeuenceClassification
from scipy.special import softmax
import tqdm
```

Define tokenizer and model

```python
Model = f"cardiffnlp/twitter-roberta-base-sentiment-latest"
tokenizer = AutoTokenizer.from_pretrained(Model)
model = AutoModelForSeuenceClassification.from_pretrained(Model)
```

Evaluating on an example

```python
example = "I love programming in Python!"
token = tokenizer(example, return_tensors='pt')
output = model(**token)
score = output[0][0].detach().numpy()
score = softmax(score)
```
